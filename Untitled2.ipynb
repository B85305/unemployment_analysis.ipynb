```python
import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("Unemployment.csv")

df.head()
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df.dropna(inplace=True)
```

```python
df.isnull().sum()
```

```python
df.describe()
```

```python
print(df.columns)
```

```python
df['DATE'] = pd.to_datetime(df['DATE'])
```

```python
df.info()
```

```python
plt.figure(figsize=(12,6))

plt.plot(
    df['DATE'],
    df['SLUEM1524ZSIND']
)

plt.title("Unemployment Trend")
plt.xlabel("Date")
plt.ylabel("Unemployment Rate")

plt.xticks(rotation=90)

plt.show()
```

```python
March 2020
```

```python
covid_data = df[
    df['DATE'] >= '2020-03-01'
]
```

```python
covid_data.head()
```

```python
plt.figure(figsize=(12,6))

plt.plot(
    covid_data['DATE'],
    covid_data['SLUEM1524ZSIND']
)

plt.title("Covid Impact on Unemployment")

plt.xticks(rotation=90)

plt.show()
```

```python
region_data = df.groupby(
    'DATE'
)['SLUEM1524ZSIND'].mean()

print(region_data)
```

```python
plt.figure(figsize=(10,6))

sns.heatmap(
    df.corr(numeric_only=True),
    annot=True
)

plt.show()
```

```python
df['Month'] = df['DATE'].dt.month
```

```python
df.head()
```

```python
monthly = df.groupby('Month')['SLUEM1524ZSIND'].mean()

print(monthly)
```

```python
monthly.plot(
    kind='line',
    marker='o'
)

plt.title(
    "Monthly Unemployment Trend"
)

plt.show()
```
